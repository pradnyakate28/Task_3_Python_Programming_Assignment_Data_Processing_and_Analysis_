```python
# Problem 1
# Finding employees eligible for highest performance bonus

employees = {
    "Pradnya": 95,
    "Harshali": 88,
    "Rohan": 95,
    "Suresh": 84
}

# Find maximum score
max_score = max(employees.values())

# Store top performers
top_list = []

for emp in employees:
    if employees[emp] == max_score:
        top_list.append(emp)

print("Top Performers Eligible for Bonus:")
print(", ".join(top_list), "(Score:", max_score, ")")
```

```python
# Problem 2
# Counting repeated keywords in search query

query = "Buy mobile phone buy phone online"

# Convert to lowercase
query = query.lower()

# Remove punctuation manually
clean_text = ""
for ch in query:
    if ch.isalnum() or ch == " ":
        clean_text += ch

words = clean_text.split()

frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

# Display only repeated words
result = {}

for word in frequency:
    if frequency[word] > 1:
        result[word] = frequency[word]

print(result)
```

```python
# Problem 3
# Store only even sensor readings

sensor_readings = [3, 4, 7, 8, 10, 12, 5]

valid_data = []

for i in range(len(sensor_readings)):
    if sensor_readings[i] % 2 == 0:
        valid_data.append((i, sensor_readings[i]))

print("Valid Sensor Readings (Hour, Value):")
print(valid_data)
```

```python
# Problem 4
# Counting domain usage percentage

emails = [
    "pradnya@gmail.com",
    "harshali@yahoo.com",
    "rohan@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]

domain_data = {}

for email in emails:
    parts = email.split("@")
    domain = parts[1]

    if domain in domain_data:
        domain_data[domain] += 1
    else:
        domain_data[domain] = 1

total = len(emails)

for d in domain_data:
    percent = (domain_data[d] / total) * 100
    print(d, ":", int(percent), "%")
```

```python
# Problem 5
# Detect sales more than 30% above average

sales = [1200, 1500, 900, 2200, 1400, 3000]

average = sum(sales) / len(sales)

limit = average + (0.30 * average)

for day in range(len(sales)):
    if sales[day] > limit:
        print("Day", day + 1, ":", sales[day])
```

```python
# Problem 6
# Detect duplicate user IDs

user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]

count_ids = {}

for uid in user_ids:
    if uid in count_ids:
        count_ids[uid] += 1
    else:
        count_ids[uid] = 1

for uid in count_ids:
    if count_ids[uid] > 1:
        print(uid, "→", count_ids[uid], "times")
```
